```python
import pandas as pd
```

```python
cars = pd.read_csv("cars.csv")
```

```python
display(cars)
```

# **A. POSITIONAL AND LABEL-BASED SLICING**

```python
print("Shape of cars:", cars.shape)
print("Columns in cars:", list(cars.columns))
```

```python
cars_6_to_10 = cars.iloc[6:10]
cars_6_to_10
```

```python
cars_6_to_10 = cars.loc[6:10, ['Model', 'mpg', 'hp','wt']]
cars_6_to_10
```

```python
display (cars_6_to_10)
```

# **B. MODEL LOOKUP**

```python
toyota = cars[cars["Model"] == "Toyota Corolla"]
```

```python
pontiac = cars[cars["Model"] == "Pontiac Firebird"][["Model", "mpg", "hp", "wt"]]
```

```python
display(toyota)
display(pontiac)
```

# **C. MULTI-MODEL SUBSETTING**
